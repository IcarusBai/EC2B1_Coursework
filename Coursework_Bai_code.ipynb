## 2. Data Collection & Cleaning

### 2.1 Load data from Excel table

```python
data = pd.read_excel('Data/pwt110_download_2025_11_24.xlsx', sheet_name = 'Data', header = 0)
```

### 2.2 Clean data

#### 2.2.1  Construct price index for scaling

```python
# Define price index
data["price_index"] = data["cgdpe"] / data["rgdpe"]
# Calculate the price index of 2017 and 2021
P_2017 = (
    data.loc[data["year"] == 2017, ["country", "price_index"]]
    .set_index("country")
)

P_2021 = (
    data.loc[data["year"] == 2021, ["country", "price_index"]]
    .set_index("country")
)
```

#### 2.2.2 Filter data and change the base year from 2021 to 2017

```python
# Filter data by columns. We need country, year, real gdp, population, real capital, employment, human capital, investment share of GDP, labor share of GDP, export share of GDP, and import share of GDP.
data_filtered = data.loc[:, ("country", "year", "rgdpe", "pop","rnna", "emp","rgdpna","hc","csh_i","labsh", "csh_x", "csh_m")]
#Drop empty values
data_filtered = data_filtered.dropna(subset=["rgdpe", "pop"])

# Calculate real dgp per person
data_filtered["rgdpe_pc"] = data_filtered["rgdpe"] / data_filtered["pop"]

# select Chile as a country
data_Chile = data_filtered.loc[data_filtered["country"] == "Chile"].copy()
# Changing base year 2021 to 2017 of rdgpe, rnna, and rgdpe_pc
scale = (
    P_2021.loc["Chile", "price_index"] /
    P_2017.loc["Chile", "price_index"]
)

cols_to_rebase = ["rgdpe", "rnna","rgdpe_pc","rgdpna"]

for col in cols_to_rebase:
    data_Chile[col] = data_Chile[col]*scale
# Reset the index
data_Chile = data_Chile.reset_index(drop = True)
```

#### 2.2.3 Specify the time period we wish to analyse

```python
# Subset the RGDP per capita series
ymax = 2023
ymin = 1951 # Starting at 1951 because data for Chile in 1950 is unavailable.

# We use logical indexing to subset the data to rows of years ymin <= year <= ymax
y = data_Chile.loc[np.logical_and(data_Chile["year"] <= ymax, data_Chile["year"] >= ymin), "rgdpe_pc"]
Y = data_Chile.loc[np.logical_and(data_Chile["year"] <= ymax, data_Chile["year"] >= ymin), "rgdpe"]
lgy = np.log(y)
lgY = np.log(Y)
data_Chile = data_Chile[data_Chile["year"] >= ymin] # we keep only years after ymin

# Compute separate sample sizes for the subsample used for estimation and the whole sample.
T = len(Y) # sample size used for regression
T_all = data_Chile["year"].max() - (ymin - 1) # number of all years in the data after ymin
```

## 3. Trend Calculation

Justification of using exponential linear model for rGDP and rGDPpc trend regression:
- We focus on log-linear (exponential) trends because long-run growth is best interpreted in terms of growth rates rather than levels; additive specifications impose implausible long-run dynamics for Chile, and higher-order log trends provide no additional insight beyond a linear specification.

```python
# First regressor x1 is T x 1 vector of ones, second regressor x2 is the vector 1, 2, ..., T, third regressor x3 is vector 1, 4, 9, ..., T^2.
# The dependent variable is per capita GDP in *levels*

x1 = np.empty(T) # initialise an empty vector for the first regressor
x2 = np.empty(T) # initialise an empty vector for the second regressor
x3 = np.empty(T) #initialise an empty vector for the third regressor

for t in range(T):
    x1[t] = 1.
    x2[t] = t + 1 # recall that Python starts indexing at 0
    x3[t] = (t + 1)**2
```

### 3.1 The trend of real GDP per capita

```python
# Model: log(yt) = a + bt
a_exp_lin, b_exp_lin = get_regression_coefs(lgy, x1, x2)
```

```python
# Initialise predicted values yhat
lgyhat_exp_lin = np.empty(T_all)

# Create loop to compute trend for all years
for t in range(T_all):
    lgyhat_exp_lin[t] = a_exp_lin + b_exp_lin * (t + 1)
```

### 3.2 The trend of real GDP

```python
# Model: log(Yt) = a + bt
a_exp_lin, b_exp_lin = get_regression_coefs(lgY, x1, x2)
```

```python
# Initialise predicted values yhat
lgYhat_exp_lin = np.empty(T_all)

# Create loop to compute trend for all years
for t in range(T_all):
    lgYhat_exp_lin[t] = a_exp_lin + b_exp_lin * (t + 1)
```

## 4. Extract TFP Measure

To extract a measure of total factor productivity (TFP), we assume a neutral Cobb–Douglas production function augmented with human capital with constant returns to scale. Output is produced using physical capital, labour, and human capital:

$$
Y_t = A_t K_t^{\alpha} (h_t L_t)^{1-\alpha}
$$

where:
- $Y_t$ is real GDP at constant 2017 national prices (`rgdpna`)
- $K_t$ is the real capital stock (`rnna`)
- $L_t$ is total employment (`emp`)
- $h_t$ is human capital index, based on years of schooling and returns to education (`hc`)
- $A_t$ is TFP

Following the benchmark value suggested in the project guidelines, we set $\alpha = 0.3$ (and later check robustness to alternative values). Taking natural logarithms, TFP can be obtained as a residual:

$$
\ln A_t
= \ln Y_t
- \alpha \ln K_t
- (1-\alpha) \ln h_t
- (1-\alpha)\ln L_t 
$$

This method interprets TFP as the component of output not explained by observable inputs, capturing long-run changes in productivity such as efficiency gains and technological progress. The resulting TFP series is used in subsequent analysis (growth accounting and the comparison between TFP and labour productivity).


### 4.1 Calculate TFP series

```python
# Define capital share
alpha = 0.3
# Calculate log values
data_Chile['ln_Y'] = np.log(data_Chile['rgdpna'])
data_Chile['ln_k'] = np.log(data_Chile['rnna'])
data_Chile['ln_l'] = np.log(data_Chile['emp'])
data_Chile['ln_hc'] = np.log(data_Chile['hc'])
# Calculate log TFP using the Cobb-Douglas production function
data_Chile['ln_tfp'] = (
    data_Chile['ln_Y']
    - alpha * data_Chile['ln_k']
    - (1 - alpha) * data_Chile['ln_l']
    - (1 - alpha) * data_Chile['ln_hc']
)

data_Chile['tfp'] = np.exp(data_Chile['ln_tfp'])
```

### 4.2 Plot for TFP over time

```python
# Plotting TFP over time
sns.set_theme('talk', style='whitegrid')

years = data_Chile['year']
tfp = data_Chile['tfp']
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(years, tfp, label='TFP (level)', linewidth=2)
ax.set_xlabel('Year')
ax.set_ylabel('TFP')
ax.set_title('TFP over time — Chile')
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()
```

### 4.3 Check robustness of TFP calculation to capital share assumption

```python
# Calculating ln_tfp and tfp for alpha = 0.25 and 0.35
alphas = {'a025': 0.25, 'a035': 0.35}
for key, alpha_val in alphas.items():
    ln_col = f'ln_tfp_{key}'
    level_col = f'tfp_{key}'
    if ln_col not in data_Chile.columns or level_col not in data_Chile.columns:
        data_Chile[ln_col] = (
            data_Chile['ln_Y']
            - alpha_val * data_Chile['ln_k']
            - (1 - alpha_val) * data_Chile['ln_l']
            - (1 - alpha_val) * data_Chile['ln_hc']
        )
        data_Chile[level_col] = np.exp(data_Chile[ln_col])

# Plotting TFP for different alpha values
years = data_Chile.index
fig, axes = plt.subplots(figsize=(14, 8))

lw = 2.5
axes.plot(years, data_Chile['tfp'], label='TFP (alpha=0.3)', linewidth=lw, color='C0')
axes.plot(years, data_Chile['tfp_a025'], label='TFP (alpha=0.25)', linewidth=lw, color='C1')
axes.plot(years, data_Chile['tfp_a035'], label='TFP (alpha=0.35)', linewidth=lw, color='C2')
axes.set_title('Figure: TFP: alpha=0.25, 0.3, 0.35')
axes.set_xlabel('Year')
axes.set_ylabel('TFP (level)')
axes.legend()
axes.grid(True)

plt.tight_layout()
plt.show()
```

- Although the level of TFP mechanically changes with α, the long-run trend and turning points remain almost identical across α = 0.25, 0.3 and 0.35, supporting the robustness of our productivity analysis.

### 5.2 Plot log rGDP and rGDP pc with 4 subperiods on background over time

```python
# Plot with colored subperiods and per-period fitted lines

periods_plot = [
    (1951, 1971, 'lightblue',   '1951–1971'),
    (1972, 1988, 'lightyellow', '1972–1988'),
    (1989, 1999, 'lightgreen',  '1989–1999'),
    (2000, 2019, 'lightcoral',  '2000–2019'),
]

lw = 3
fit_lw = 2.5

fig, axes = plt.subplots(1, 2, figsize=(18, 7), sharey=False)

years_arr = data_Chile["year"].values
log_rgdpe = np.log(data_Chile['rgdpe'].values)
log_rgdpe_pc = np.log(data_Chile['rgdpe_pc'].values)

series_info = [
    (log_rgdpe,    'log real GDP',            'ln(rGDP)',            'Panel A: Log real GDP'),
    (log_rgdpe_pc, 'log real GDP per capita',  'ln(rGDP per capita)', 'Panel B: Log real GDP per capita'),
]

for ax, (series, label, ylabel, title) in zip(axes, series_info):
    # Background shading
    for start, end, color, plabel in periods_plot:
        ax.axvspan(start, end, alpha=0.3, color=color, label=plabel)

    # Actual data
    ax.plot(years_arr, series, color='black', linewidth=lw, label=label)

    # Per-period OLS fitted line
    for start, end, color, plabel in periods_plot:
        mask = (years_arr >= start) & (years_arr <= end)
        x_sub = years_arr[mask]
        y_sub = series[mask]
        if len(x_sub) < 2:
            continue
        coeffs = np.polyfit(x_sub, y_sub, 1)
        fitted = np.polyval(coeffs, x_sub)
        g = coeffs[0] * 100  # annualised growth rate %
        ax.plot(x_sub, fitted, linewidth=fit_lw, linestyle='--',
                label=f'{plabel} trend ({g:+.2f}% p.a.)')

    ax.set_ylabel(ylabel)
    ax.set_xlabel('Year')
    ax.set_title(title)
    ax.legend(fontsize=8, loc='upper left')

fig.suptitle('Figure 2: Four subperiods present dinstinct growth rates')
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

### 5.3 Perform growth accounting for each subperiod

```python
# Define growth accounting function
def growth_accounting(df, start, end, alpha):
    gY = df.loc[end, "ln_Y"] - df.loc[start, "ln_Y"]
    gk = df.loc[end, "ln_k"] - df.loc[start, "ln_k"]
    gl = df.loc[end, "ln_l"] - df.loc[start, "ln_l"]
    gA = df.loc[end, "ln_tfp"] - df.loc[start, "ln_tfp"]
    gh = df.loc[end, "ln_hc"] - df.loc[start, "ln_hc"] # Calculate growth rate of each factor

    cap = alpha * gk
    lab = (1 - alpha) * gl
    hc  = (1 - alpha) * gh # multiply growth of capital, labor and hc by alpha or 1- alpha to get contribution

    return {
        "Total GDP growth": f"{gY:.2f}",
        "TFP contribution": f"{gA:.2f} ({gA/gY*100:.2f}%)",
        "Capital contribution": f"{cap:.2f} ({cap/gY*100:.2f}%)",
        "Labour contribution": f"{lab:.2f} ({lab/gY*100:.2f}%)",
        "Human capital contribution": f"{hc:.2f} ({hc/gY*100:.2f}%)",
    }
```

```python
# Perform growth accounting for each period
alpha = 0.3  # Reset alpha to baseline value
results = []
data_Chile = data_Chile.set_index("year")
# Loop through each period and calculate growth accounting
for (s, e) in periods:
    res = growth_accounting(data_Chile, s, e, alpha)
    res["Period"] = f"{s}-{e}"
    results.append(res)

# Append the decomposition result for the whole period
res = growth_accounting(data_Chile, periods[0][0], periods[-1][1], alpha)
res["Period"] = f"{periods[0][0]}-{periods[-1][1]}"
results.append(res)

ga_table = pd.DataFrame(results)
ga_table = pd.DataFrame(results).set_index("Period")
ga_table
```

## 8. Trade Openness and TFP

We examine the descriptive relationship between trade openness and total factor productivity in Chile. Trade openness is defined as:

$$\text{openness}_t = \text{csh\_x}_t + \text{csh\_m}_t$$

where `csh_x` is the export share of GDP and `csh_m` is the import share (reporte

### 8.1 Calculating trade openness

```python
# Define trade openness as export share + |import share|
# In PWT, csh_m is negative, so csh_x + csh_m gives exports minus imports;
# for a standard openness measure we use csh_x + abs(csh_m).
data_Chile["openness"] = ((abs(data_Chile["csh_x"]) + abs(data_Chile["csh_m"]))) 

# Take logs
data_Chile["log_openness"] = np.log(data_Chile["openness"])
```

```python
data_Chile["csh_x"].describe()
```

### 8.2 Regression & visualisation of openness and TFP

```python
# OLS with robust (HC1) standard errors
valid = data_Chile[['log_openness', 'ln_tfp']].dropna()
X = sm.add_constant(valid['log_openness'])
y = valid['ln_tfp']

ols_model = sm.OLS(y, X).fit(cov_type='HC1')

# Print regression table
print(ols_model.summary2(title='OLS: ln(TFP) on ln(Openness) — HC1 robust SEs'))

# Plot log TFP and log openness: dual-axis time series + scatter
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Panel A: Dual-axis time series
ax1 = axes[0]
color_tfp = '#4C72B0'
color_open = '#DD8452'

ln1, = ax1.plot(data_Chile['year'], data_Chile['ln_tfp'],
                color=color_tfp, linewidth=2.5, label='log TFP')
ax1.set_xlabel('Year')
ax1.set_ylabel('log TFP', color=color_tfp)
ax1.tick_params(axis='y', labelcolor=color_tfp)

ax2 = ax1.twinx()
ln2, = ax2.plot(data_Chile['year'], data_Chile['log_openness'],
                color=color_open, linewidth=2.5, label='log Openness')
ax2.set_ylabel('log Openness', color=color_open)
ax2.tick_params(axis='y', labelcolor=color_open)

lines = [ln1, ln2]
ax1.legend(lines, [l.get_label() for l in lines], loc='upper left')
ax1.set_title('Panel A: Log TFP and Log Trade Openness over time')
ax1.grid(True, alpha=0.3)

# Panel B: Scatter plot with robust OLS line 
ax3 = axes[1]
ax3.scatter(valid['log_openness'], valid['ln_tfp'],
            alpha=0.6, edgecolors='white', s=60, color='#55A868')

# Plot fitted line from robust OLS
x_fit = np.linspace(valid['log_openness'].min(), valid['log_openness'].max(), 100)
X_fit = sm.add_constant(x_fit)
y_fit = ols_model.predict(X_fit)
ax3.plot(x_fit, y_fit, color='black', linewidth=2, linestyle='--',
         label=f'OLS fit (β = {ols_model.params.iloc[1]:.2f}, robust SE = {ols_model.bse.iloc[1]:.2f})')

corr = valid['log_openness'].corr(valid['ln_tfp'])
ax3.set_xlabel('log Openness')
ax3.set_ylabel('log TFP')
ax3.set_title(f'Panel B: Correlation (r = {corr:.3f})')
ax3.legend(fontsize=9)
ax3.grid(True, alpha=0.3)

fig.suptitle('Figure 9: Openness is correlated with TFP')
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
# Plot log TFP and labor productivity (GDP per worker): dual-axis time series
fig, ax1 = plt.subplots(figsize=(10, 6))
color_tfp = '#4C72B0'
color_lp = '#DD8452'

ln1, = ax1.plot(data_Chile['year'], data_Chile['tfp'],
                color=color_tfp, linewidth=2.5, label='TFP')
ax1.set_xlabel('Year')
ax1.set_ylabel('TFP', color=color_tfp)
ax1.tick_params(axis='y', labelcolor=color_tfp)

ax2 = ax1.twinx()
ln2, = ax2.plot(data_Chile['year'], data_Chile['labour_prod'],
                color=color_lp, linewidth=2.5, label='Labour productivity')
ax2.set_ylabel('GDP per worker', color=color_lp)
ax2.tick_params(axis='y', labelcolor=color_lp)

ax1.legend([ln1, ln2], [ln1.get_label(), ln2.get_label()], loc='upper left')
ax1.set_title('Figure 7: Labor productivity and TFP present similar upwards trend')
ax1.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 9.2 How labor productivity and TFP are correlated

```python
# Scatter plot of log TFP vs log labor productivity
fig, ax = plt.subplots(figsize=(8, 6))

valid = data_Chile[['ln_labour_prod', 'ln_tfp']].dropna()

ax.scatter(valid['ln_labour_prod'], valid['ln_tfp'],
           alpha=0.6, edgecolors='white', s=60, color='#55A868')


corr = valid['ln_labour_prod'].corr(valid['ln_tfp'])
ax.set_xlabel('log Labour Productivity')
ax.set_ylabel('log TFP')
ax.set_title(f'Figure 8: Labor productivity and TFP are highly correlated')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
